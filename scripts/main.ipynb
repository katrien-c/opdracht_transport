## 6. Route-specifieke Analyse

Selecteer hieronder een specifieke route om gedetailleerde analyse te zien.

```python
# Maak een dropdown widget voor route selectie
unique_routes = sorted(df['route_id'].unique().tolist())

route_dropdown = widgets.Dropdown(
    options=unique_routes,
    value=unique_routes[0] if unique_routes else None,
    description='Route:',
    disabled=False,
    style={'description_width': 'initial'}
)

# Output widget voor de plots
output = widgets.Output()

def on_route_change(change):
    """Callback functie die wordt aangeroepen wanneer de route selectie verandert"""
    with output:
        output.clear_output(wait=True)
        selected_route = change['new']
        print(f"Geselecteerde route: {selected_route}\n")
        
        # Genereer de plots voor de geselecteerde route
        plot_total_tasks_per_day_per_route(df, selected_route)
        plot_total_tasks_per_day_per_route_from_begin_and_end(df, selected_route)

# Koppel de callback aan de dropdown
route_dropdown.observe(on_route_change, names='value')

# Toon de widget en output
display(widgets.VBox([
    widgets.HTML("<h3>Selecteer een Route voor Gedetailleerde Analyse</h3>"),
    route_dropdown,
    output
]))

# Trigger initial plot
on_route_change({'new': route_dropdown.value})
```
